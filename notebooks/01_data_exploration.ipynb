# TripMind: Data Exploration & Feature Analysis
Comprehensive exploratory data analysis (EDA) of Indian destination metadata, attractions catalog, and expense records.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_dest = pd.read_csv('../data/processed/destinations.csv')
print("Total Destinations:", len(df_dest))
df_dest.head()
```

```python
print("Distribution across categories:")
print(df_dest['category'].value_counts())

df_dest['category'].value_counts().plot(kind='bar', color='#06b6d4', figsize=(10, 4))
plt.title('Destination Distribution by Category')
plt.ylabel('Count')
plt.show()
```

```python
print("Average Daily Budget summary:")
print(df_dest['average_budget'].describe())
```
