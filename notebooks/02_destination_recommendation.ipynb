# Model 1: Destination Recommendation Engine
Content-based filtering using cosine similarity, TF-IDF vectorization, and weighted multi-attribute matching.

```python
import sys; sys.path.append('..')
import pandas as pd
from backend.app.ml.recommendation_model import DestinationRecommendationModel
from backend.app.ml.train_recommendation import evaluate_recommendations

df = pd.read_csv('../data/processed/destinations.csv')
model = DestinationRecommendationModel()
model.fit(df)

metrics = evaluate_recommendations(model, df, top_k=5)
print("Recommendation Metrics:", metrics)
```

```python
# Example recommendation inference
ranked = model.rank(user_interests=['Mountains', 'Adventure'], budget=20000, duration_days=4, travellers_count=2, top_k=5)
pd.DataFrame(ranked)[['name', 'state', 'category', 'match_percentage', 'estimated_budget']]
```
